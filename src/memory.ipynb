  - model：会话摘要时要使用的模型
  - trigger：会话摘要的触发时机，有三种设置：
    - fraction (float): 模型上下文大小的比例（0-1）
    - tokens (int): 令牌数量
    - messages (int): 消息数量
  - keep：是指触发摘要后要保留的消息
    - fraction (float): 要保留的消息占模型上下文大小的比例（0-1）
    - tokens (int): 要保留的消息的令牌数量
    - messages (int): 要保留的消息数量

```python
from langchain.agents import create_agent
from langchain.agents.middleware import SummarizationMiddleware
from langgraph.checkpoint.memory import InMemorySaver
from langchain_core.runnables import RunnableConfig


# 初始化checkpointer
checkpointer = InMemorySaver()
# 初始化中间件
middleware = SummarizationMiddleware(
    model="deepseek-chat",
    trigger=("messages", 3), #  触发时机，当消息数超过3时，进行总结
    keep=("messages", 1) #  保留的会话数，超过2条
)
```

```python
agent_middleware = create_agent(
    model="deepseek-chat",
    middleware=[middleware],
    checkpointer=checkpointer,
)

config: RunnableConfig = {"configurable": {"thread_id": "1"}}
# 制造长会话历史
agent_middleware.invoke({"messages": "你好，我是小c."}, config)
agent_middleware.invoke({"messages": "我最喜欢的运动是篮球"}, config)
agent_middleware.invoke({"messages": "我最喜欢的动物是猫猫"}, config)
# 测试效果
final_response = agent_middleware.invoke({"messages": "你还记得我吗？"}, config)


for message in final_response["messages"]:
    message.pretty_print()
```

```python
import sqlite3
import json
from pprint import pprint

from langgraph.checkpoint.sqlite import SqliteSaver

# 1) 连接数据库
conn = sqlite3.connect("res/checkpoint.db", check_same_thread=False)
conn.row_factory = sqlite3.Row
cursor = conn.cursor()

# 2) 初始化 checkpointer
checkpointer = SqliteSaver(conn)
checkpointer.setup()

# 3) 创建 agent
agent = create_agent(model="deepseek-chat", checkpointer=checkpointer)

# 4) 调用，写入 thread_2
config_2 = {"configurable": {"thread_id": "thread_2"}}
response_2 = agent.invoke(
    {"messages": [HumanMessage(content="你好，我叫kk，我最喜欢猫猫。")]},
    config_2
)

print("=== agent response ===")
print(response_2)

# 5) 手动提交，确保外部数据库工具能看到
conn.commit()

# 6) 打印数据库里有哪些表
print("\n=== tables ===")
cursor.execute("SELECT name FROM sqlite_master WHERE type='table' ORDER BY name")
for row in cursor.fetchall():
    print(row["name"])

# 7) 查询 thread_2 的所有落库记录
print("\n=== checkpoints rows for thread_2 ===")
cursor.execute(
    """
    SELECT
        checkpoint_id,
        checkpoint_ns,
        thread_id,
        parent_checkpoint_id,
        type,
        metadata,
        checkpoint
    FROM checkpoints
    WHERE thread_id = ?
    ORDER BY rowid
    """,
    ("thread_2",)
)

rows = cursor.fetchall()
print(f"rows count = {len(rows)}")

for i, row in enumerate(rows, start=1):
    print(f"\n--- row {i} ---")
    print("checkpoint_id:", row["checkpoint_id"])
    print("checkpoint_ns:", row["checkpoint_ns"])
    print("thread_id:", row["thread_id"])
    print("parent_checkpoint_id:", row["parent_checkpoint_id"])
    print("type:", row["type"])

    # metadata 可能是 JSON 字符串，也可能是 bytes
    metadata = row["metadata"]
    print("metadata:", metadata)

    # checkpoint 通常是二进制或序列化内容，先原样打印类型
    checkpoint = row["checkpoint"]
    print("checkpoint type:", type(checkpoint))
    print("checkpoint raw:", checkpoint)

# 8) 如果你想只看 thread_2 的最关键字段
print("\n=== simplified view ===")
cursor.execute(
    """
    SELECT thread_id, checkpoint_id, checkpoint_ns, parent_checkpoint_id, type
    FROM checkpoints
    WHERE thread_id = ?
    ORDER BY rowid
    """,
    ("thread_2",)
)
for row in cursor.fetchall():
    pprint(dict(row))
```
